#### Multiple Subplot(다중 플롯)

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
ax1 = plt.axes()
ax2 = plt.axes([0.65, 0.5, 0.2, 0.3])
```

```python
for i in range(1, 10):
    plt.subplot(3, 3, i)    # (행의수, 열의 수, 인덱스)
    plt.text(0.5, 0.5, str((3,3,i)), ha='center')
```

```python
fig = plt.figure()
fig.subplots_adjust(hspace=0.4, wspace=0.7)

for i in range(1, 10):
    plt.subplot(3, 3, i)
    plt.text(0.5, 0.5, str((3,3,i)), ha='center')
```

```python
fig, ax = plt.subplots(3, 3, sharex='col', sharey='row')
```

```python
grid = plt.GridSpec(2, 3, wspace=0.4, hspace=0.4)

plt.subplot(grid[0,0])
plt.subplot(grid[0,1])
plt.subplot(grid[0,2])
plt.subplot(grid[1,0])
plt.subplot(grid[1,1])
plt.subplot(grid[1,2])
```

```python
grid = plt.GridSpec(2, 3, wspace=0.4, hspace=0.4)

plt.subplot(grid[0,0])
plt.subplot(grid[0,1:])

plt.subplot(grid[1,:2])
plt.subplot(grid[1,2])
```

```python
plt.figure(figsize=(5,6))

x = range(1, 21)
columns = [np.random.randn(20) * i for i in range(1, 7)]

i = 0
for c in columns:
    i += 1

    plt.subplot(3,2,i)
    plt.plot(x, c, '-o')
    plt.xlim(-1, 21)
    plt.ylim(c.min()-1, c.max()+1)
```
