```python
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-darkgrid')
```

1. 다음 데이터를 이용하여 학생의 5일간 공부 시간을 선 그래프로<br>
day = [1, 2, 3, 4, 5]<br>
study = [2, 3, 5, 4, 6]<br><br>

* 요구사항<br>
선 색상은 파란색으로 지정<br>
선 스타일은 실선으로 지정<br>
마커는 원형으로 지정<br>
제목은 Study Hours로 지정<br>
x축은 Day, y축은 Hours로 지정

```python
day = [1, 2, 3, 4, 5]
study = [2, 3, 5, 4, 6]

plt.plot(day, study, 'b-o')
plt.title('Study Hours')
plt.xlabel('Day')
plt.ylabel('Hours')
```

2. 다음 데이터를 이용하여 일주일간의 기온 변화를 나타내기<br>
day = [1, 2, 3, 4, 5, 6, 7]<br>
temp = [20, 22, 21, 25, 27, 26, 29]<br><br>

* 요구사항<br>
선 색상은 빨간색으로 지정<br>
선 스타일은 점선으로 지정<br>
마커는 사각형으로 지정<br>
선 두께는 2로 지정<br>
마커 크기는 8로 지정<br>
x축 범위는 1~7, y축 범위는 15~35로 설정<br>
제목은 Weekly Temperature로 지정<br>

```python
day = [1, 2, 3, 4, 5, 6, 7]
temp = [20, 22, 21, 25, 27, 26, 29]

plt.plot(day, temp, 'r--s', linewidth=2, markersize=8)
# plt.xlim(1, 7)
# plt.ylim(15, 35)
plt.axis([1,7,15,35])
plt.title('Weekly Temperature')
```

3. 두 학생의 시험 점수를 하나의 그래프에서 비교하기<br>
exam = [1, 2, 3, 4, 5]<br><br>
student_A = [70, 75, 80, 85, 90]<br>
student_B = [65, 72, 78, 82, 88]<br><br>

* 요구사항<br>
두 학생의 점수를 하나의 그래프에 표시<br>
A 학생은 파란색 실선, 원형 마커를 사용<br>
B 학생은 초록색 점선, 삼각형 마커를 사용<br>
범례에는 Student A, Student B를 표시<br>
범례 위치는 오른쪽 아래로 지정<br>
y축 범위는 60~100으로 지정<br>
제목은 Exam Score Comparison으로 지정

```python
exam = [1, 2, 3, 4, 5]
student_A = [70, 75, 80, 85, 90]
student_B = [65, 72, 78, 82, 88]

plt.plot(exam, student_A, 'b-o', label='Student A')
plt.plot(exam, student_B, 'g--^', label='Student B')
plt.title('Exam Score Comparison')
plt.xlabel('Exam')
plt.ylabel('Score')

plt.ylim(60, 100)
plt.legend(loc='lower right')
```

4. 다음 데이터를 이용하여 월별 매출 변화를 그래프로<br>
    month = [1, 2, 3, 4, 5, 6]<br>
    sales = [120, 150, 130, 180, 200, 230]<br><br>

* 요구사항<br>
그래프의 제목과 축 이름을 한글로 작성<br>
선 색상은 보라색으로 지정<br>
선 스타일은 일점쇄선으로 지정<br>
마커는 다이아몬드로 지정<br>
x축 범위는 1~6, y축 범위는 100~250으로 지정<br>
제목 글자 크기는 16, x축과 y축 글자 크기는 12로 지정<br>
범례에는 매출액을 표시

```python
plt.rcParams['font.family'] = 'Malgun Gothic'

month = [1, 2, 3, 4, 5, 6]
sales = [120, 150, 130, 180, 200, 230]

plt.plot(month, sales, '-.D', color='purple', label='매출액')
plt.title('월별 매출 변환', fontsize=16)
plt.xlabel('월', fontsize=12)
plt.xlabel('매출액', fontsize=12)

plt.axis([1,6,100,250])
plt.legend()
```

5. 다음 데이터를 이용하여 세 제품의 6개월간 판매량을 비교하는 그래프를 작성<br>
month = [1, 2, 3, 4, 5, 6]<br><br>
product_A = [100, 120, 140, 130, 160, 180]<br>
product_B = [90, 110, 115, 140, 150, 170]<br>
product_C = [80, 95, 105, 120, 135, 150]<br><br>

* 요구사항<br>
세 제품의 판매량을 하나의 그래프에 나타내기<br>

    - 제품별 스타일을 다음과 같이 지정<br>

        | 제품 | 색상 | 선 스타일	| 마커 |
        | :---: | :---: | :---: | :---: |
        | A | red	| 실선 | 원형 |
        | B | blue	| 점선 | 사각형 |
        | C | green	| 점 | 삼각형 |

    - 세 그래프 모두 선 두께를 2로 지정<br>
    - x축 범위를 1~6, y축 범위를 50~200으로 지정<br>
    - 제목은 Product Sales Comparison, 글자 크기는 18로 지정<br>
    - x축 이름은 Month, y축 이름은 Sales로 지정<br>
    - 범례를 왼쪽 위에 표시<br>
    - x축 눈금은 1부터 6까지 표시

```python
month = [1, 2, 3, 4, 5, 6]
product_A = [100, 120, 140, 130, 160, 180]
product_B = [90, 110, 115, 140, 150, 170]
product_C = [80, 95, 105, 120, 135, 150]

plt.plot(month, product_A, 'r-o', linewidth=2, label='Product A')
plt.plot(month, product_B, 'b--s', linewidth=2, label='Product B')
plt.plot(month, product_C, 'g:^', linewidth=2, label='Product C')

plt.axis([1,6,50,200])
plt.title('Product Sales Comparison', fontsize=18)
plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks([1,2,3,4,5,6])
plt.legend(loc='upper left')
```
