#### Scatter Plot(산점도)
 : 두 변수의 관계를 2차원 평면위에 점으로 나타내는 그래프
- 상관관계 파악
    - 양의 상관관계 : x가 증가하면 y도 증가하는 경향
    - 음의 상관관계 : x가 증가하면 y는 감소하는 경향
    - 무상관 : x와 y사이의 특별한 선형 관계가 없음
- 이상치 감지를 한눈에 파악할 수 있다
- 점들이 특정 영역에 뭉쳐(군집(cluster)) 있는지 확인이 쉽다

- 마커 문자 목록

| 마커문자 | 모양 | 설명 |
| :---: | :---: | :---: |
| . | point | 점 |
| , | pixel | 점(픽셀의 크기) |
| o | circle | 원 |
| v | triangle_down | 아래쪽 삼각형 | 
| ^ | triangle_up | 위쪽 삼각형 |
| < | triangle_left | 왼쪽 삼각형 |
| > | triangle_right | 오른쪽 삼각형 | 
| 1 | tri_down | Y형 |
| 2 | tri_up | Y형을 꺼꾸로 |
| 3 | tri_left | Y형을 왼쪽방향 | 
| 4 | tri_right | Y형을 오른쪽방향 |
| s | square | 사각형 |
| p | pantagon | 오각형 |
| * | star | 별 | 
| h / H | hexagon | 육각형 |
| + | plus | 더하기 |
| x | x | x표시 | 
| D | diamond| 다이아몬드 |
| d | thin diamond | 얇은 다이아몬드 |
| \| | vline | 수직선 |
| _ | hline | 수평선 |


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.figure(figsize=(8,4))
markers = ['.',',','o','v','^','<','>','1','2','3','4','s','p','*','h','H','+','x','D','d','|','_']

for m in markers:
    plt.plot(np.random.rand(5), np.random.rand(5), m, label="'{0}'".format(m))

plt.legend(loc='center right', ncol=2)
plt.xlim(0, 1.5);
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.scatter(x, y, marker='o');
# plt.plot(x, y, 'o')
```

```python
for i in range(9):
    x = np.arange(1000)
    y = np.random.randn(1000).cumsum()
    plt.scatter(x, y, alpha=0.2, c=y, cmap='viridis')
```

#### 버블 챠트

```python
x = np.random.randn(100)
y = np.random.randn(100)
colors = np.random.rand(100)
sizes = np.random.rand(100) * 1000

plt.scatter(x,y, c=colors, s=sizes, alpha=0.3, cmap='magma')
plt.colorbar();
```

```python
x = np.random.randn(200)
y1 = np.random.randn(len(x))
y2 = 1.1 * np.exp(x)

ax1 = plt.plot()
plt.scatter(x, y1, color='indigo', alpha=0.3, label='no correlation')
plt.scatter(x, y2, color='blue', alpha=0.3, label='correlation')
plt.grid(True)
plt.legend();
```
