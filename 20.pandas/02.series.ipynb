#### Series
: 인덱스가 붙어있는 1차원 데이터

```python
from pandas import Series
import numpy as np
```

```python
# data1 = Series([45, 73, 13, 48])
data1 = Series([45, 73, 13, 48, '더조은', True, 5.384])
data1
```

```python
data2 = Series(np.random.randint(1,10,size=(5,)), index=['1회', '2회', '3회', '4회', '5회'])
data2
```

```python
data2['2회'] = 9
data2
```

```python
data2.index
```

```python
data2.values
```

```python
age = {'kim':28, 'lee':32, 'park':25}
data3 = Series(age)
print(data3)

data4 = Series(age, index=['no1', 'no2', 'no3'])
data4
```
