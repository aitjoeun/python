```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(np.ones((4,4)), index=['a','b','c','d'])
df
```

```python
# b행의 1열에 결측치 삽입
df.loc['b', 1] = np.nan
df

df.loc['b':, 3] = np.nan
df
```

```python
pd.isnull(df)
```

```python
pd.notnull(df)
```

```python
# 3열의 True의 합계(count의 의미)
df[3].isnull().sum()
```

```python
# 한행에 결측치가 하나라도 있으면 제외
df.dropna()
```

```python
# 한행이 모두 결측치 일때만 제외
df.dropna(how='all')
```

```python
# 결측값을 임의의 값으로 대체
df.fillna(0)
```

```python
# 1열의 모든 결측치는 5로, 3열의 모든 결측치는 7로 대체
df.fillna({1:5, 3:7})
```

```python
df.fillna('unknown')
```
